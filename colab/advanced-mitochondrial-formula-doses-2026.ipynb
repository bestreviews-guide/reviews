## A dose audit, not a review

This post is not a verdict on whether the formula is good or bad. It is a dose audit: I took the doses printed on the label and compared them with the doses used in the studies those ingredients come from. A formula can have the right ingredients and the wrong amounts, and the only way to tell is to put the two numbers side by side. The label is honest - every dose is printed, no proprietary blend, no hiding. That honesty is exactly what makes the audit possible, and it is also what makes the results uncomfortable.

## CoQ10: 50 mg against 100 to 300 mg

CoQ10 is the headline ingredient, dosed at 50 mg. The trials that showed real effects - the ones the sales page is built on - used 100 to 300 mg a day. So the bottle gives you between a sixth and a half of the study dose, depending on which trial you pick. CoQ10 is not cheap, and that is almost certainly why the dose is what it is: at 200 mg a day, this bottle would cost considerably more. But the gap matters. If you buy this formula expecting the CoQ10 effect from the trials, you are getting a fraction of the amount that produced it - and the sales page does not say that out loud.

## PQQ: 5 mg against 20 mg

PQQ is the second key ingredient, dosed at 5 mg. The studies that put PQQ on the map used 20 mg a day - four times what is in the bottle. This one is harder to explain away, because PQQ is dosed in milligrams either way and the cost difference between 5 and 20 mg is small. The 5 mg dose is common in commercial formulas, and it is not nothing. But the headline studies - the ones that showed the mitochondrial effects everyone quotes - used 20 mg. A quarter of the study dose is a quarter of the study dose, and no marketing language changes the arithmetic.

## D-ribose: 250 mg against 15 grams

D-ribose is the third ingredient, and the gap here is the widest of the three: the bottle contains 250 mg, while the research doses run closer to 15 grams a day - fifteen thousand milligrams, sixty times what is in the bottle. To be fair, 15 grams is a large dose, and nobody expects a tablet formula to carry that much - the tablets would be enormous. But that is exactly the point: D-ribose at 250 mg is a garnish, not a dose. It is on the label because it looks good there, not because 250 mg reproduces the research. The number on the label and the number in the studies are in different leagues, and a buyer who does not check will never know.

## The rest of the eleven

The same pattern holds across the rest of the list: real ingredients, printed doses, amounts below the study levels. This is a formula built to a price, not to the research doses - nothing dishonest about that, since the doses are printed, but it is a choice the buyer should understand: you are buying a well-chosen list at modest amounts, not the study protocols themselves.
