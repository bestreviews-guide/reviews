## What DL185 and myHMB actually are

DL185 is a leucine metabolite blend built around the amino acid leucine, the trigger that switches muscle protein synthesis on. Leucine is the signal that tells your muscles to build and repair, and older muscles need a bigger signal than younger ones do. myHMB is a branded form of HMB, a compound your body makes from leucine, which is sold for slowing muscle breakdown. So the two ingredients work from opposite sides: one pushes muscle building up, the other pushes muscle breakdown down. That is a sensible pairing, and it is aimed squarely at the problem people over 50 actually have — not building more muscle for the gym, but keeping the muscle they already own.
