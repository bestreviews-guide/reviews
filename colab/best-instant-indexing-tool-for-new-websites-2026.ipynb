## The three doors into Google's index

Google finds a new site through one of three doors, and you should open all three. The first door is a sitemap: you submit it in Google Search Console, which is free and takes about ten minutes to set up, and Google reads the list of your pages. The second door is a link from a site Google already visits - a profile, a directory listing, a guest post, anything with a crawlable link pointing at you. The third door is a direct request: Search Console lets you ask for indexing on a specific URL, about ten requests a day. Open all three on day one, and the first crawl usually comes within a day or two. Open only one, and you are waiting on a single door that Google may not walk through for a while. A good tool does not replace these doors. It adds a fourth: bulk discovery for everything the free doors cannot reach.

## What it cannot do - and why I say so

Here is what no indexing tool can do for a new site, and I would rather you hear it from me than learn it from a refund request. Nobody can force Google to index a page, and nobody can promise you a date. Indexing is Google's decision, full stop. The Google Indexing API cannot be used for arbitrary customer URLs either - it is built for job postings and live streams, not for your blog or your client's site, so any tool claiming API-powered instant indexing for any URL is not telling you the whole truth. And an indexed page is not a ranked page: indexing means Google has found you, ranking means Google thinks you deserve to be seen, and those are two very different jobs. If a tool promises you a rank or a date, walk away.
