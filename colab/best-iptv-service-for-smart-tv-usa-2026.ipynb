## What your internet needs to handle on a big screen

A big screen shows every flaw in a stream, so your internet matters more on a smart TV than anywhere else. Streams run in HD, Full HD and 4K depending on the channel and your connection, and a 4K stream on a 55-inch TV needs a stable 25 Mbps or better. If your internet struggles with 4K on YouTube, it will struggle with IPTV in 4K too. Before you buy a yearly plan, test on the exact TV and the exact connection you will use every day - not on a phone, not on fast Wi-Fi somewhere else. If your Wi-Fi is weak in the living room, a wired connection to the TV or a better router position will do more for your viewing than any subscription upgrade. IPTV lives or dies on your connection, and the big screen is where that shows.
