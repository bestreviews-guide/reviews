## The five things that decide it

First, coverage for your region. A service built for US viewers should cover the USA properly, and if your household watches UK, Canadian or European channels too, it should cover those as well. Broad coverage matters more than a big number: a service claiming tens of thousands of channels is selling a number, not a lineup. What matters is whether the channels you watch every week are there and stable. Second, stream quality. Streams run in HD, Full HD and 4K depending on the channel and your connection, and a 4K stream needs a stable 25 Mbps or better. The best service is the one whose streams hold up on your connection, not the one with the biggest quality claim on its sales page. Third, connections. One to four connections means the whole household watches on different screens under one subscription - the living-room TV, a bedroom TV, a phone, a tablet. A service that charges per screen gets expensive fast. Fourth, price with a monthly option. The best service lets you test it for a month before asking you to commit to a year. Fifth, support you can reach - an email address and a messaging channel, with answers that come back fast and clear.
