## The system, in plain English

The Encyclopedia of Power Foods teaches a Traditional Chinese Medicine way of eating. The framework is the Five Elements - Wood, Fire, Earth, Metal, Water - each tied to an organ system and a food colour. The idea is that eating foods matched to the elements keeps your body in balance: green foods for the Wood element and the liver, red foods for Fire and the heart, yellow for Earth and the spleen, white for Metal and the lungs, black for Water and the kidneys. The book adds herbal wisdom - superfoods for Jing, Qi and Shen, TCM herbs, adaptogens - plus seasonal eating, mindful eating, and bonus chapters on Tai Chi, Qi Gong and mindfulness. As a framework, it is memorable and easy to use. That is its real strength.

## What the system gets right

The best thing about this system is that it makes variety easy. "Eat five colours" is a rule anyone can follow at a supermarket, and varied plant foods are consistently linked to better health in real research. Seasonal eating is sensible too - food in season tastes better, costs less, and varies naturally across the year. Mindful eating, slowing down, paying attention to meals: also sensible, and also supported by research. None of this requires believing in the Five Elements as literal medicine. It works as a practical framework for eating better, and that is enough. The book's core advice - eat a colourful variety of whole foods, cut back on ultra-processed stuff - is advice I would give anyone.

## Where it stops being a guide

The problem is the story the system arrives in. The sales page is told by Connor Kennedy, a Chicago police officer, whose wife Lucy collapsed at their daughter's wedding - her lab results, he says, showed a drug interaction. That sent him to a remote mountain village in China called Bama Yao, where he says people live well past 100 in robust health without modern medicine. The system he brings back is called "Oriental Optimization". None of this is verifiable. No trial has tested this book's system. The longevity claims - a decade of extra life, disease reversal - are the kind of claims that need evidence, and the page offers a story instead. I am not saying the village does not exist. I am saying a story is not evidence, and a $37 ebook should not need a tall tale to sell a sensible idea.

## What you cannot see before you pay

Here is a practical complaint: you buy this book blind. No page count, no sample chapter, no contents list. For a book, that is a strange way to sell - bookshops let you read the first pages for a reason. The 30-day refund softens it: no-quibble, full refund, and you keep the book. But you should not need a refund to find out whether a book is what you wanted. The refund is generous, and I credit the seller for it, but the missing preview is still a weakness.

## Price and value

$37 one-time, shown discounted from $97, no subscription, no hidden charges, instant download. That is cheap for a reference book, and the keep-the-book refund removes most of the risk. If you read it in a weekend and keep two or three ideas - eat five colours, eat seasonally, slow down at meals - you have got your $37 of value. If you bought it for the longevity promise, you will be disappointed, because no book delivers a decade of extra life.

## Who should buy it, and who should not

Buy it if you like food-as-medicine frameworks and want a memorable way to eat more varied whole foods. Buy it if $37 feels fair for a reference book you will dip into for ideas. Skip it if you want a meal plan with recipes - this is a book you read, not a plan you follow. Skip it if you need proof before you believe a health claim. And skip it if the Bama Yao story is what convinced you, because the story is the weakest part of the product, not the strongest.

## My final verdict

3.1 out of 5. The system is a useful way to eat better. The story selling it is not. Buy it for the food ideas and the price - not for the promise of a decade of extra life.
