## My verdict

3.1 out of 5. The Encyclopedia of Power Foods is a $37 ebook from 365 Daily Health that teaches a Traditional Chinese Medicine way of eating - five elements, five organs, five food colours - with a 30-day refund and instant download. The core idea underneath it is genuinely sensible: eat a colourful variety of whole foods and cut back on ultra-processed stuff, and your health will thank you. But the book arrives wrapped in a dramatic tale about a remote Chinese village where people live past 110, plus disease-reversal stories I cannot verify, and you buy without seeing a page count, a sample chapter or a contents list. Buy it for the food ideas and the price - not for the promise of a decade of extra life.

## What is The Encyclopedia of Power Foods?

First, the thing to get straight: this is a book, not a bottle. It is a downloadable ebook from 365 Daily Health, compiled by its founders Kenny and Wen Li. The sales page is told by Connor Kennedy, a Chicago police officer, who opens with his wife Lucy collapsing at their daughter's wedding - her lab results, he says, showed the collapse was triggered by an interaction between two drugs she was taking. That sent him looking for a natural way to restore her health, and eventually to a remote mountain village in China called Bama Yao, where he says people live well past 100 in robust health without modern medicine. The system he brings back is called "Oriental Optimization" - Lucy named it that, according to the page. Its core is Traditional Chinese Medicine: five elements (Wood, Fire, Earth, Metal, Water), each tied to an organ and a food colour.

## What is inside the book

The book organises "Power Foods" by the Five Elements, each tied to an organ system and a food colour. It covers the healing power of food colours, advanced nutritional strategies and herbal wisdom - superfoods for Jing, Qi and Shen, TCM herbs, adaptogens - and fitting TCM into modern life through seasonal eating and mindful eating. There are also bonus chapters on Tai Chi, Qi Gong and mindfulness. As a reference book about food, that is a reasonable table of contents. The problem is what you cannot see before you pay: no page count, no sample chapter, no contents list. You are buying a book blind, which is a strange way to sell a book.

## The food advice: what is actually sound

Strip away the story and the core advice is sensible: eat a colourful variety of whole foods, match foods to seasons, cut back on ultra-processed products, and pay attention to how you eat, not just what you eat. Real research backs the general direction - diets rich in varied plant foods are consistently linked to better health outcomes. The five-colours idea is a practical way to make sure your plate is varied: red foods, green foods, yellow foods, white foods, black foods. You do not need Traditional Chinese Medicine to justify eating a rainbow, but the framework makes it easy to remember, and easy to remember is valuable.

## The tall tale: what costs it the score

The Bama Yao story is the problem. A remote village where people live past 110 in robust health without modern medicine, disease-reversal stories, a system that supposedly adds a decade of life - none of it is verifiable, and none of it has been tested. No trial has tested this book's system. The story is marketing dressed as a documentary, and it is doing the heavy lifting for a $37 ebook. I am not saying the village does not exist or that its people are not healthy. I am saying a longevity claim that dramatic needs evidence, and the sales page offers a story instead. That is what costs the book its score - not the food advice, which is fine, but the tale wrapped around it.

## Price and guarantee

$37 one-time, shown discounted from $97. No subscription is mentioned, and the seller states there are no hidden charges or monthly billings. It is a digital download, delivered in seconds. The 30-day guarantee is no-quibble, full refund by email or letter - and you keep the book. That is a fair deal: $37 is cheap for a reference book, and keeping the book after a refund removes most of the risk. My advice: buy it for the food ideas, read it in a weekend, and keep what is useful.

## Who should buy it, and who should not

Buy it if you like food-as-medicine ideas, you want a structured way to eat more varied whole foods, and $37 feels fair for a reference book with a keep-the-book refund. Skip it if you are buying it for the longevity promise - a decade of extra life is not something a $37 ebook can deliver. Skip it if you want a meal plan with recipes, because this is a reference book you read, not a plan you follow day by day. And skip it if you need proof before you believe a health claim, because this book offers a story where proof should be.

## My final verdict

3.1 out of 5. Good food sense inside a tall tale. Buy it for the food ideas and the price - not for the promise of a decade of extra life.
