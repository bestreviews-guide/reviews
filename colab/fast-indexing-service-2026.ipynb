## Before anything goes out: the checklist

I never submit a list I have not checked first, because a broken URL wastes a credit and a blocked page wastes a submission. My checklist is short and I run it every time. One: every URL returns a proper 200 response - I click a random sample, and if one fails, I check the whole list. Two: no noindex tag hiding in the code, usually left over from staging. Three: robots.txt is not blocking the pages. Four: the content is real - not a duplicate, not a thin placeholder, not a page I would be embarrassed for Google to read. Five: the list is one URL per line, no typos, no trailing spaces. Ten minutes of checking saves a batch of wasted credits, and it is the least glamorous part of the job, which is exactly why most people skip it.

## The list itself

A hundred URLs sounds like a lot until you collect them. A new site launch gives you forty pages in an afternoon. A backlink campaign gives you fifty fresh links in a week. A round of profile creation gives you twenty more. I keep a running text file and add URLs to it as they go live, so by Friday I usually have a batch ready. The file has one URL per line and nothing else - no titles, no notes, no formatting. When the list hits a hundred, it is time to submit. Smaller lists work too, but batching is the point: one submission, one check the next day, instead of fifty little chores spread across the week.

## The afternoon after

Here is the part people get wrong: after submitting, I stop. I do not refresh Search Console every ten minutes. I do not resubmit the same URLs. I do not buy a second tool to double up. The signals are out, Googlebot will follow within hours, and the indexing decision takes hours to days after that. Resubmitting does not make Google move faster - it just burns credits on URLs that are already in the queue. So I close the tab and do something else. The waiting is the job now, and the job takes a day or two, not an afternoon.

## The next morning: the check

The next morning, I check. In Search Console, I paste a sample of the URLs into URL Inspection and read the answers. "URL is on Google" means it made it. "Discovered - currently not indexed" means Google knows about it but has not crawled it yet - those get resubmitted in the next batch. "Crawled - currently not indexed" means Google read the page and declined it, and the fix is on the page itself, not in another submission. I note which ones are stuck, fix what I can fix, and the stuck ones go back into the list for the next round. That loop - submit, wait, check, fix, resubmit - is the whole workflow, and it is the only one that actually gets pages indexed.
