## What herbal remedies can and cannot do

Let me set the frame honestly, because herbal remedy books live in a grey area. Some herbs have real research behind them: ginger for nausea, peppermint for digestion, chamomile for sleep. Others have centuries of traditional use and very little modern evidence.  A good herbal book tells you all three things: what works, what is traditional but unproven, and what to avoid. A bad one only tells you the good news. The value of a bundle like this is not that it contains secret knowledge.
