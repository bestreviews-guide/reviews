## The short answer

For a new page on an established site that Google already visits daily, indexing usually takes hours to a few days. For a new page on a brand-new site, expect days to weeks. For a full new website, the homepage often appears within a few days, while deeper pages can take two to four weeks to all show up. Those are normal ranges, not worst cases. If your page is outside them, something is slowing it down, and the rest of this notebook is about finding what.

## Why the range is so wide

Indexing time depends on three things. First, crawl frequency: how often Googlebot visits your site, which depends on your site's age, authority, and publishing rhythm. Second, discovery: how quickly Google hears the page exists, which depends on links, sitemaps, and submissions. Third, the page itself: clean, useful pages get indexed faster than thin or duplicated ones. A new site scores badly on all three at once, which is why everything feels slow at the start. An established site with daily publishing scores well on all three, which is why its pages appear almost immediately. Same Google, same process, totally different wait.

## New websites: days to weeks

A brand-new domain starts at zero. Googlebot has no history with it and no reason to visit often. The homepage usually gets crawled within a few days of the first submission or the first backlink, because that is when Google first hears about the domain. Deeper pages follow over the next one to three weeks as the bot builds a crawl pattern. You can shorten this by launching with internal links already in place, submitting your sitemap on day one, requesting indexing for the homepage, and getting even one or two backlinks early. What you cannot do is skip the queue entirely. New sites wait. That is the tax on being new.

## Established sites: hours to days

Once a site has history, the timeline collapses. Googlebot already visits every day, sometimes several times a day, so a new page gets discovered almost as soon as it is linked internally. Crawling follows within hours, and indexing usually lands within a day or two. This is why big publishers seem to rank minutes after publishing: it is not favoritism, it is crawl frequency earned over years. Your site gets there by publishing consistently and keeping its technical setup clean, so the bot keeps coming back.

## What slows it down

Several things quietly add days or weeks. A page with no internal links pointing at it can sit undiscovered for a long time. A sitemap that is out of date or never submitted leaves Google guessing. Thin content - a few generic sentences - can get crawled and then declined, which looks exactly like "not indexed" from the outside. Duplicate content does the same. A slow server makes Googlebot deprioritise your pages. And a robots.txt block or a noindex tag stops everything cold; I have seen sites wait a month over a leftover staging block that took thirty seconds to remove.

## How to speed it up

Work the three levers in order. Discovery: link the page internally from a crawled page, submit your sitemap, and request indexing in Search Console for the pages that matter most - roughly ten a day. Crawl frequency: publish regularly and keep the site fast and clean, so the bot's visits get more frequent over time. Page quality: give Google something worth indexing, specific and different from your other pages.
