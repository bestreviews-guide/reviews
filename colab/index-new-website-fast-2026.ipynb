## Why new sites wait

Google discovers new sites through links and sitemaps. A brand-new domain has neither - no links pointing at it, no history, no crawl budget. So it sits in a queue behind millions of other URLs. The November core update made this worse: during a rollout, Google deprioritises new and light pages. I have seen fresh domains take three weeks for a first crawl, and some pages never get crawled at all. Waiting is not a strategy. It is a hope.

## Step one: the basics, done properly

Before you submit anything, get the foundations right. A sitemap with every URL you want found, submitted in Search Console. A robots file that does not block anything by accident - I have seen more sites killed by a stray disallow than by any penalty. Every page returning 200, with a real title and real content, not a placeholder. Internal links from the homepage to every important page, because Google follows links and your own links count. None of this is exciting. All of it is required.

## Step two: submit the same day

The day a page goes live, submit it. Not next week, not when you remember - the same day. Search Console gives you about ten requests a day, account-wide, which covers a small site. For a bigger launch, use a bulk indexing tool and paste up to 100 URLs per request. Submit the homepage first, then the money pages, then everything else.

## Step three: build links that point at it

A page with no inbound links is a page Google has no reason to visit. That is why the first backlinks matter so much for a new site. A Telegra.ph page, a JustPaste post, a Gist, a Notion page, a profile link - each one is a road Google can follow to your site. Build five or ten of them in the first week, all pointing at your homepage and your first money pages. Then submit those backlink URLs too, because a backlink Google has not indexed is a road nobody has driven on.

## Step four: check and re-submit

Forty-eight hours after submitting, check. Search Console's URL inspection tool tells you whether a URL is indexed, discovered but not indexed, or not found at all. Re-submit anything still undiscovered. Repeat weekly until the important pages are in. This is the boring part and the part most people skip, which is why most new sites stay invisible. The loop is the work: publish, submit, check, re-submit.

One more thing that speeds discovery up: put your new URLs in your existing pages. If you already run another site, a single link from an indexed page on the old site to the new site's homepage is the fastest discovery signal there is. Google follows links, and a link from a page it already crawls is worth more than any submission. Even a link from a social profile or a directory listing helps, because it gives the crawler a road to follow.
