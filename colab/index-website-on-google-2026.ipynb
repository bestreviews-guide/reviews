## The three steps, every time

Every page on your site goes through the same three steps. First, discovery: Google has to find out the page exists. That happens through links from pages Google already knows, through your sitemap, or through a direct submission. Second, crawling: Googlebot fetches the page and reads the content, the links, the structure. Third, indexing: Google analyses what it read and decides whether the page earns a place in the index. If any step stalls, the page stays invisible. Most new-site owners obsess over step three, but step one is where most pages actually get stuck.

## Step 1: Search Console, on day one

Before anything else, add your site to Google Search Console. It is free, it takes about ten minutes, and it is the only place where Google talks to you directly about your site. Verify ownership, then look at the dashboard. Search Console is where you submit sitemaps, request indexing, and see exactly which pages Google has and which ones it is ignoring, with reasons. Skip this step and you are working blind.

## Step 2: Sitemap and your first requests

A sitemap is a simple list of your pages that tells Google what exists and what matters. Most platforms generate one automatically - usually at yourdomain.com/sitemap.xml. Submit it in Search Console under Sitemaps, and keep it updated whenever you publish. Then use the URL Inspection tool to request indexing for your homepage and your most important pages first.

## Step 3: Link everything internally

A page with no links pointing at it is a page Google may never find. Every new page should be linked from somewhere Google already crawls: your homepage, a popular post, a category page. Internal links are how Googlebot travels through your site, and they are also how Google judges which pages matter. A page linked from your homepage gets crawled sooner than a page buried three clicks deep with no incoming links. This step costs nothing and it is the one people skip most often. Before you publish anything new, decide where it will be linked from. That one habit alone fixes most slow-indexing problems on small sites.

## Step 4: Get a few real backlinks

One backlink from a page Google visits daily can pull Googlebot to your site within hours. You do not need hundreds. A couple of genuine links - a profile, a directory listing, a guest mention - give Google a second doorway into your site. This is also why new sites feel so slow: with zero backlinks, Google has no reason to visit often. Even a handful of links changes the crawl pattern noticeably within a week or two.

## What slows everything down

A few things quietly sabotage the routine. A noindex tag left on a live page, often left over from staging. A robots.txt file blocking everything, also usually a staging leftover. Slow pages or errors, which make Googlebot deprioritise your site. Thin content that gets crawled and then declined - no submission tool will change Google's mind until the page itself improves. Fix the blockers once, follow the steps in order, and give it a few days.
