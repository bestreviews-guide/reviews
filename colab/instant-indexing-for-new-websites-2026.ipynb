## Day 0: your site is invisible

Let me be blunt about where a new website starts: nowhere. Google has never visited it, has no record of it, and has no reason to come looking. You can type your domain into Google on launch day and get nothing, and that is normal. It is not a penalty, it is not a bug, and it is not something you did wrong. It is the starting line. Every site began exactly here. The ones that got found fast did a few specific things in the first week, and that is what this post is about.

## Week 1: what Google does first

In the first week, Google's job is discovery, and it finds your site through one of three doors: a sitemap you submitted, a link from a site it already visits, or a direct request through Search Console. Until one of those doors opens, nothing happens at all. Add the site to Google Search Console - it is free, it takes about ten minutes - and submit your sitemap. Then request indexing for your homepage. That single request is the fastest free lever you have, and it usually gets the homepage crawled within a day or two. After that, Google follows the links on your homepage to the rest of your site, which is why internal links matter from day one - a page with no links pointing at it might never get found at all.

## The discovery problem nobody warns you about

Here is the part that catches new owners off guard. Search Console gives you roughly ten indexing requests a day. A forty-page launch needs four days just for the requests, and that is before you count the days you forget. Worse, half of what a new site needs found is not on the site at all - the backlinks and profiles around the launch - and there is no Search Console for pages you do not own. So the manual route breaks the moment your list gets long, and the pages you leave waiting can sit undiscovered for weeks. That is the discovery problem in one sentence: the free method does not scale, and a new site is the one place where scale matters most.

## What to submit first - and what to hold back

Not every page deserves the same urgency. Submit your homepage first, then your two or three most important pages. After that, submit in batches: the rest of your site pages, then the backlink pages, then the profiles. Do not submit pages that are not ready: a noindex tag left over from staging, a broken image, or placeholder text will get crawled and declined, and you burned a request on a page Google was right to refuse. Check each batch before it goes out: every URL returns a proper 200, nothing is blocked, the content is real. Ten minutes of checking saves a week of wondering why nothing worked.

## The honest timeline for a new site

Here is the timeline, without the marketing. Homepage submitted through Search Console: crawled within a day or two, indexed within a few days after that. Inner pages linked from the homepage: a few days behind. Pages with no internal links and no backlinks: a week or more, sometimes much longer. Backlink pages on small sites: two to four weeks without a push, days with one. None of those numbers are instant, and all of them are normal. "Instant" in indexing honestly means fast submission, never guaranteed indexing - your URLs go out in minutes, Googlebot usually follows within hours, and the indexing decision takes hours to days after that. Indexing is Google's decision, full stop. Anyone who tells you a new site will be fully indexed in fifteen minutes is selling you the first step and calling it the whole journey.
