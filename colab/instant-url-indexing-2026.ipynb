## The one-URL job

Sometimes you have exactly one URL that matters. A new homepage. A money page you just rewrote. A post that has to be live in search by Monday. For that job, the single-URL route is the right tool, and it is free. Open Google Search Console, paste the URL into URL Inspection, and request indexing. Google usually crawls it within a day. That is the fastest free lever you have, and for one URL, nothing beats it - not because it is fast, but because it is direct. You are asking Google yourself, about your own page, through Google's own tool. There is no middleman, no queue, no credit. One URL, one request, done.

## The hundred-URL job

Now flip it. You launched a new site with forty pages. You finished a backlink campaign with fifty fresh links. You created twenty profiles in a week. The single-URL route breaks immediately, because Search Console gives you roughly ten requests a day. A hundred URLs take ten days of clicking, and that is before you count the days you forget. Worse, half of those URLs are backlinks on other people's sites, and there is no Search Console for those at all - you cannot request indexing for a page you do not own. The hundred-URL job needs a different tool, because the problem is no longer speed. It is scale.

## The math that decides

Here is the line, in plain numbers. Under ten URLs a week, all on your own site: use Search Console, free, one at a time. Over ten URLs a week, or any URLs you do not own: use a bulk tool. A bulk submission takes a hundred URLs in one request and ten minutes of work. At that volume, the manual route is not just slow - it is the thing stopping you from doing the work at all. People do not skip indexing because they are lazy. They skip it because the manual method does not fit the size of the job, and nobody told them there was another method.

## What bulk submission actually does

A bulk tool does one job: it sends fast discovery signals for your whole list at once. You paste up to 100 URLs, pick a mode, and submit. The signals go out in minutes - pings to Google's discovery systems, IndexNow-style notifications, sitemap pings, links from pages Google already visits. Googlebot usually follows within hours, and indexing takes hours to days after that. That is the whole mechanism, and it is the same three steps as any indexing: discovery, crawl, decision. The bulk tool only speeds up the first step. It does not, and no tool does, speed up Google's decision. When someone sells you "instant URL indexing," they are selling you the fast first step and calling it the whole journey.

## The mistakes people make with bulk

Three mistakes, all common. First, submitting URLs that are not ready: broken pages, noindex tags, thin content. A bulk tool will send signals for them, Google will crawl them, and Google will decline them. You burned credits on pages that were never going to pass. Second, resubmitting the same list every day. The URLs are already in the queue - resubmitting does not move them forward, it just costs you more credits. Third, never checking. Submitting without checking is like mailing letters and never asking if they arrived. Check in Search Console two days after every batch, note what moved and what did not, and fix the stuck ones before you resubmit.

One URL: Search Console, free, done. A hundred URLs: bulk submission, one request, ten minutes.
