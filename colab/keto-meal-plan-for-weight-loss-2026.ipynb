## Why keto helps people lose weight

Keto works by cutting carbs hard enough that your body switches to burning fat for fuel. That switch does two useful things. First, it usually kills your appetite — fat and protein keep you full far longer than bread and pasta do, so most people eat fewer calories without counting them. Second, it removes the foods people overeat most: snacks, sweets, takeaway carbs. That is the honest mechanism. It is not magic, and it is not a faster metabolism. It is appetite control plus fewer temptations, and for a lot of people that is exactly what they needed. The research backs the short term well. The long term is harder, because staying low-carb for a year is a different job from staying low-carb for a month.
