## What nerve pain actually feels like

It is not like muscle pain. Muscle pain is dull, aching, and it gets better with rest. Nerve pain is tingling, burning, pins and needles, numbness that comes and goes, and it is often worse at night.  It is also the kind of pain that gets worse slowly and quietly, which is why people put up with it for years. And it is the kind of pain that makes sleep hard, which makes everything else worse. A supplement cannot fix damaged nerves. What it can do, at best, is support the environment around them and help you sleep through the worst of it.
