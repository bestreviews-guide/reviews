## Why gums bleed

Bleeding when you floss is not usually about brushing too hard. It is gingivitis - inflammation caused by the bacteria living along the gum line. Your mouth is full of bacteria, good and bad, and when the bad side wins, the gums get inflamed, puffy and quick to bleed. Brushing and flossing remove the film the bacteria live in, which is why dentists push them so hard. But they do not change which bacteria live there. That is the gap a probiotic is sold into: add friendly bacteria, shift the balance, calm the inflammation. The logic is sound. The question is whether the dose matches the research.
