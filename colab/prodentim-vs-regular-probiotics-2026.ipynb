## What a gut probiotic does

A gut probiotic is a capsule you swallow, usually with strains like Lactobacillus acidophilus or Bifidobacterium longum, aimed at digestion, bloating and gut balance. It is designed to survive stomach acid and reach the intestines. That is a hard journey, and a good product is built for it - delayed-release capsules, high CFU counts, strains chosen for the gut. What it does not do is spend time in your mouth. It passes through in seconds. So if you bought a gut probiotic hoping it would help your gums, you bought the right idea for the wrong address. The bacteria never get the chance to colonise your mouth, because they were never meant to live there.
